```python
import pickle
import pandas as pd

# Load the saved model and scaler
with open("svr_bundle.pkl", "rb") as file:
    saved_objects = pickle.load(file)

model = saved_objects["model"]
scaler = saved_objects["scaler"]
scaler_output =saved_objects["scaler_output"]
columns = saved_objects["columns"]
```

```python
# New inputs in the same order used during training
new_input = pd.DataFrame(
    [[1300, 12000, 4000, 0, 1]], columns=columns)
```

```python
new_input
```

```python
# Apply the saved preprocessing
scaled_input = scaler.transform(new_input)
```

```python
scaled_input
```

```python
# Predict on the scaled output scale
scaled_prediction = saved_objects["model"].predict(scaled_input)

print(f"Predicted profit: {scaled_prediction[0]:,.2f}")
```

```python
# Convert the prediction back to the original output scale
original_prediction = saved_objects["scaler_output"].inverse_transform(
    scaled_prediction.reshape(-1, 1))

print(f"Predicted profit: {original_prediction[0, 0]:,.2f}")
```
